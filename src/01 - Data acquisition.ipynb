## Case-Study 01: Malta International Airport

Loading the [Malta International Airport](https://maltairport.com) passenger movements statistical data in different formats.

```python
# Load libraries
import pandas as pd
import os.path as osp

#P.S. pd is an alias for the pandas libarary, which allows us to load data in a data frame.

#P.S. os.path is used to have an OS agnostic code, which runs on Windows, Linux and Mac. 
```

```python
# Load the MIA dataset
df = pd.read_csv(osp.join('..', 'data', 'mia', 'miaPassengers 01.csv'))

# We are identifying the data file in an OS agnostic manner using relative positioning, relative to the current file. The .. symbol
# represents navigating to one folder up (parent folder). The osp.join() take care of adding the folder seperator based on OS.
#pd is the alias for pandas which loads the csv contents in data frame.

#Preview the data structure and do a data integrity check
df.info()
```

Either from the data dictionary or by actually previewing the data file we get an indication of the number of rows and columns, to verify with what has been loaded. Here we undertake a data integrity check and confirm that we have 19 rows, 2 columns, no nulls and two 64-bit integer variables. Next we preview the data to verify the start and end of the dataframe. 

```python
#Preview the start of the data
df.head() # the default value is 5 rows
```

```python
#Preview the end of the data
df.tail() # the default value is 5 rows
```

Following a data integrity check we confirm that the data file was consistent, that no notes were added and no aggregate information was appended to the end.

```python
# Accessing a file using file locks to get variable names
with open(osp.join('..', 'data', 'mia', 'miaPassengers.names'), 'r', encoding='utf-8-sig') as f:
    columns = f.read().strip().split(',')

columns
```

We created a pointer `f` to the file we want to access and attached a read lock. We also define the encoding, meaning the character space to use to interpret the content, utf8. The `utf-8-sig` variant is needed to support the specific BOM character of this file. Every file has a BOM and EOF character. 

When reading the contents of the file we remove whitespace using the `strip()` command, then we create a list with the contents of the file using the `split(',')` command since they are comma seperated. 

```python
df = pd.read_csv(osp.join('..', 'data', 'mia', 'miaPassengers.data'), names=columns)
df.info()
df.head()
```
